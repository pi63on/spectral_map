# Processing In the Spectrum readings

```python
dir_path = r'm3/spectra/'
files = os.listdir(dir_path)
files_cp = sorted(files, key = lambda x: int(x.split('_')[0]))
```

```python
files_cp
```

```python
plt.rcParams.update({'font.size': 16})
COLOR = 'white'
mpl.rcParams['text.color'] = COLOR
mpl.rcParams['axes.labelcolor'] = COLOR
mpl.rcParams['xtick.color'] = COLOR
mpl.rcParams['ytick.color'] = COLOR
mpl.rcParams['axes.edgecolor'] = COLOR

for i , file in enumerate(files_cp):
    print(file)
    data = pd.read_csv(dir_path + file, delimiter = '\t', skiprows=1)
    x = data['wavelength']
    y = data['intensity']

    fig, ax = plt.subplots(figsize=(10,7), facecolor = 'black')
    plt.scatter(x,y, s = 1 , color = 'white' )
    plt.plot(x,y, linewidth = 1 , color = 'white' , alpha = .1)
    ax.set_yticklabels([])

    ax.set_facecolor("black")
    plt.xlabel("wavelength (nm)")
    plt.title(file)

    
    plt.tight_layout()
    plt.savefig(f'm3/spectra_evals_nove/{i}_eval_spectrum.png')
    # plt.show()
```

```python
plt.rcParams.update({'font.size': 16})
COLOR = 'white'
mpl.rcParams['text.color'] = COLOR
mpl.rcParams['axes.labelcolor'] = COLOR
mpl.rcParams['xtick.color'] = COLOR
mpl.rcParams['ytick.color'] = COLOR
mpl.rcParams['axes.edgecolor'] = COLOR

for i , file in enumerate(files_cp):
    print(file)
    data = pd.read_csv(dir_path + file, delimiter = '\t', skiprows=1)
    x = data['wavelength']
    y = data['intensity']
    wavel = np.zeros((len(x), 3))
    for j, wasol in enumerate(x):
        if (wasol >= 380) and (wasol <=780):
            wavel[j] = np.divide(wavelength_to_rgb(wasol), 255)
            # print(j, wavel[j])
            if (j == 78):
                wavel[j] = [0.,1.,0.21176471]
    
    # break
    fig, ax = plt.subplots(figsize=(10,7), facecolor = 'black')
    
    plt.plot(x,y, linewidth = 1 , color = 'white' , alpha = .5)
    plt.scatter(x,y, s = 2, c = wavel )
    plt.bar(x = x , height = y, width=1, color = wavel)
    ax.set_yticklabels([])

    ax.set_facecolor("black")
    plt.xlabel("wavelength (nm)")
    plt.ylabel("intensity (a.u.)")

    
    plt.tight_layout()
    plt.savefig(f'm3/spektra_evals_color/{i}_eval_spectrum.png')
    plt.close()
    # plt.show()
```
