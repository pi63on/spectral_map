```python
from gpxcsv import gpxtolist
gpx_list = gpxtolist('m2\spectral_meas_2.gpx')

#if you have pandas
import pandas as pd
df = pd.DataFrame(gpx_list)
```

```python
df['time']
df = df.drop('type', axis = 1)
df = df.drop('name', axis = 1)
df = df.drop('ele', axis = 1)
```

```python
def clean_time(time):
    time = time.replace('-', '')
    time = time.replace('T', '')
    time = time.replace(':', '')
    time = time.replace('Z', '')
    return time
```

```python
df['time'] = df['time'].apply(clean_time)
```
